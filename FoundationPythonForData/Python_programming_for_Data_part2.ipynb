# Mini Challenge.

process the data in the file "ethnicity_age_data.csv"

fill the nan values for ages based on the average age with the role based on country.


```python
#@title solution

data_file = "ethnicity_age_data.csv"
# 1. read the data
ds =pd.read_csv(data_file)
# 2. extract all ethnicity called akan under the column called Ethnicity
 # hint ds[ds[______]==_____]
akan_ds = ds[ds['Ethnicity']=="Akan"]

# 3. get the average age for that group
#hint  akan_ds[___].mean()
average_age = akan_ds['Age'].mean()



# 4. fill the nan values with the average age
ds.loc[ds['Ethnicity'] == "Akan", 'Age'] = ds.loc[ds['Ethnicity'] == "Akan", 'Age'].fillna(average_age)
```
