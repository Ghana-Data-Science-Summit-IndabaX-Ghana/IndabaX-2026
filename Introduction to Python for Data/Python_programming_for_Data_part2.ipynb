The function takes these parameters:

`DataFrame.drop_duplicates(subset=None, *, keep='first', inplace=False, ignore_index=False)`

More information on these parameters can be accessed here `https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop_duplicates.html`

### 4.4. GroupBy in Pandas

In data preparation for machine learning, each row represents a unique entity (e.g., a student, customer, or observation) with specific features (columns).

Sometimes, we want to analyze or transform data based on shared characteristics.  
For example, grouping all students with the same name or all samples from the same class.

In groupby, the column you group by becomes the index of the result, not the original rows.

Pandas provides the `groupby()` function to support this.


`groupby()` is used to group rows in a dataset based on the values of one or more columns.

It follows a three-step process:
1. **Split** the data into groups based on a column (e.g., "Name")
2. **Apply** a function to each group
3. **Combine** the results into a new structure

The pie chart is hard to visualize as a group of ownership have count 1

We would clean up the code below.

We would use a special condition of Groupby as **Groupby accepts conditions**

In our cleanup we want to label all other ownership less than 2 as others.


````python
counts = df_group['Ownership'].value_counts()
````

````python
import numpy as np
import matplotlib.pyplot as plt

plt.close('all')

counts = df_group['Ownership'].value_counts()

threshold = 0.01 * counts.sum()
cleaned_data = counts.groupby(np.where(counts >= threshold, counts.index, 'Other')).sum()

cleaned_data = cleaned_data.sort_values(ascending=False)


plt.figure(figsize=(8, 8))
cleaned_data.plot(kind='pie', autopct='%1.1f%%', labels=None, startangle=140)

plt.legend(cleaned_data.index, title="Ownership", loc="upper left", bbox_to_anchor=(1, 1))
plt.title("Distribution of Ownership of Health Facilities in Ghana", weight='bold')
plt.ylabel('')
plt.show()
````
