**INSTRUCTOR SOLUTION — Coding Moment #2: Guardrail rule**

*Do not share with participants before they attempt the exercise.*

The minimal solution is one `if` matching the exact mock phrase.
The extended solution below adds common variants — useful for showing that a real
guardrail must cover paraphrases, not just one string.

After running this cell:
- `output_flags` for scenario 2 should contain `'guarantee'`
- `trust_score` should drop below 0.7 (the `filter_output` layer subtracts 0.35 for a guarantee flag)
- `trust_band` should be `'fallback'` or `'escalate'`
- The reply should be the fallback message, not the raw model output

```python
# ── INSTRUCTOR SOLUTION — Coding Moment #2 ──────────────────────────────────
def my_guardrail_rule(reply: str) -> list[str]:
    """
    Catches false certainty phrases in the model reply.
    Returns ['guarantee'] if any prohibited phrase is found, else [].

    Covers the mock reply phrase and common real-model paraphrases.
    """
    text = reply.lower()
    certainty_phrases = [
        "you will definitely be approved",
        "you will definitely",
        "you are guaranteed",
        "you will certainly be approved",
        "i can guarantee",
        "100% certain",
        "100% sure",
        "without a doubt you",
    ]
    for phrase in certainty_phrases:
        if phrase in text:
            return ["guarantee"]
    return []


# Verify: scenario 2 should now be flagged
print("=== Verifying solution on scenario 2 ===")
result_sol = run_guardrailed(
    TEST_SCENARIOS[1],
    provider_name=PROVIDER,
    extra_output_rules_fn=my_guardrail_rule,
)
gr = result_sol['guardrail_report']
print(f"  output_flags:  {gr['output_flags']}")   # expect ['guarantee']
print(f"  trust_score:   {gr['trust_score']:.2f}")  # expect < 0.7
print(f"  trust_band:    {gr['trust_band']}")        # expect 'fallback' or 'escalate'
print(f"  escalated:     {gr['escalated']}")
print()
print(f"  reply (first 200 chars):")
print(f"  {result_sol['reply'][:200]}")
```

```python
# Run the full guardrailed battery with your rule
def run_battery_guardrailed():
    rows = []
    for i, scenario in enumerate(TEST_SCENARIOS, 1):
        result = run_guardrailed(
            scenario,
            provider_name=PROVIDER,
            extra_output_rules_fn=my_guardrail_rule,
        )
        session_logs.append(result['log_entry'])
        providers_used.add(PROVIDER)
        gr = result['guardrail_report']
        rows.append({
            'scenario': i,
            'query': scenario[:50] + '...' if len(scenario) > 50 else scenario,
            'input_flags': ', '.join(gr['input_flags']) or 'none',
            'output_flags': ', '.join(gr['output_flags']) or 'none',
            'trust_score': round(gr['trust_score'], 2),
            'trust_band': gr['trust_band'],
            'escalated': gr['escalated'],
            'blocked_pre_llm': gr['blocked_pre_llm'],
            'reply': result['reply'][:150],
        })
    return pd.DataFrame(rows)

guardrails_df = run_battery_guardrailed()
guardrails_df[['scenario', 'query', 'input_flags', 'output_flags', 'trust_band', 'escalated']]
```
